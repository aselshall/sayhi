```python
from PIL import Image, ImageChops
import os

def crop_whitespace(img, background_color=(255, 255, 255)):
    # Ensure image is RGB
    if img.mode != "RGB":
        img = img.convert("RGB")

    # Create a solid background image
    bg = Image.new("RGB", img.size, background_color)
    
    # Find the bounding box of non-background area
    diff = ImageChops.difference(img, bg)
    bbox = diff.getbbox()

    if bbox:
        return img.crop(bbox)
    else:
        return img  # No cropping needed

# Process all .tif/.tiff files in the current folder
for filename in os.listdir("."):
    if filename.lower().endswith((".tif", ".tiff")):
        print(f"Processing: {filename}")
        try:
            with Image.open(filename) as img:
                cropped = crop_whitespace(img)
                new_filename = os.path.splitext(filename)[0] + ".png"
                cropped.save(new_filename, format="PNG")
                print(f"✅ Saved cropped PNG: {new_filename}")
        except Exception as e:
            print(f"❌ Error with {filename}: {e}")
```
